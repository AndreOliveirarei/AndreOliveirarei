```python
-- Criação das tabelas principais de usuários e empréstimos
CREATE TABLE Usuarios (
    id_usuario INTEGER PRIMARY KEY AUTOINCREMENT,
    nome TEXT NOT NULL,
    cpf TEXT NOT NULL UNIQUE,
    endereco TEXT,
    telefone TEXT,
    email TEXT,
    numero_matricula TEXT NOT NULL UNIQUE,
    tipo_usuario TEXT CHECK(tipo_usuario IN ('estudante', 'professor', 'comunidade')) NOT NULL,
    status_usuario TEXT CHECK(status_usuario IN ('ativo', 'suspenso', 'bloqueado')) NOT NULL DEFAULT 'ativo'
);

CREATE TABLE Livros (
    id_livro INTEGER PRIMARY KEY AUTOINCREMENT,
    titulo TEXT NOT NULL,
    autor TEXT NOT NULL,
    editora TEXT NOT NULL,
    ano_publicacao INTEGER NOT NULL,
    edicao TEXT NOT NULL,
    numero_paginas INTEGER NOT NULL,
    quantidade_exemplares INTEGER DEFAULT 0,
    localizacao TEXT
);

CREATE TABLE Emprestimos (
    id_emprestimo INTEGER PRIMARY KEY AUTOINCREMENT,
    id_usuario INTEGER,
    id_livro INTEGER,
    data_emprestimo DATE NOT NULL,
    data_devolucao DATE,  -- Permitindo NULL para devolução
    data_prevista_devolucao DATE NOT NULL,
    renovacoes INTEGER DEFAULT 0,
    FOREIGN KEY (id_usuario) REFERENCES Usuarios(id_usuario),
    FOREIGN KEY (id_livro) REFERENCES Livros(id_livro)
);

-- Criação de Índices para os campos frequentemente consultados
CREATE INDEX idx_usuario_nome ON Usuarios(nome);
CREATE INDEX idx_livro_titulo ON Livros(titulo);
CREATE INDEX idx_emprestimo_usuario ON Emprestimos(id_usuario);
CREATE INDEX idx_emprestimo_livro ON Emprestimos(id_livro);

-- Trigger para controle de empréstimos e devoluções
-- Atualizando a quantidade de exemplares de um livro quando ocorre um empréstimo
CREATE TRIGGER atualizar_quantidade_exemplares_emprestimo
AFTER INSERT ON Emprestimos
FOR EACH ROW
BEGIN
    UPDATE Livros SET quantidade_exemplares = quantidade_exemplares - 1 WHERE id_livro = NEW.id_livro;
END;

-- Atualizando a quantidade de exemplares de um livro quando ocorre uma devolução
CREATE TRIGGER atualizar_quantidade_exemplares_devolucao
AFTER DELETE ON Emprestimos
FOR EACH ROW
BEGIN
    UPDATE Livros SET quantidade_exemplares = quantidade_exemplares + 1 WHERE id_livro = OLD.id_livro;
END;

-- Inserção de livros com 3 exemplares cada
INSERT INTO Livros (titulo, autor, editora, ano_publicacao, edicao, numero_paginas, quantidade_exemplares, localizacao)
VALUES ('O Grande Livro Da Mitologia Egípcia', 'Blanc Claudio', 'Editora XYZ', 2015, '1ª Edição', 350, 3, 'Seção A - Prateleira 1');

INSERT INTO Livros (titulo, autor, editora, ano_publicacao, edicao, numero_paginas, quantidade_exemplares, localizacao)
VALUES ('Os Segredos Do Império Romano', 'Walter Fernandes', 'Editora ABC', 2018, '1ª Edição', 400, 3, 'Seção B - Prateleira 2');

-- Inserção de usuários (Carlos Dias e Mariele Dias)
INSERT INTO Usuarios (nome, cpf, endereco, telefone, email, numero_matricula, tipo_usuario, status_usuario)
VALUES ('Carlos Dias', '111.222.333-44', 'Rua X, 123', '(11) 98765-4321', 'carlosdias@example.com', '202304', 'estudante', 'ativo');

INSERT INTO Usuarios (nome, cpf, endereco, telefone, email, numero_matricula, tipo_usuario, status_usuario)
VALUES ('Mariele Dias', '555.666.777-88', 'Rua Y, 456', '(11) 98765-8765', 'marieledias@example.com', '202305', 'professor', 'ativo');

-- Inserção de empréstimos (com devolução prevista em 7 dias)
-- Como exemplo, vamos registrar um empréstimo para cada usuário
INSERT INTO Emprestimos (id_usuario, id_livro, data_emprestimo, data_prevista_devolucao, renovacoes)
VALUES (1, 1, '2024-11-20', '2024-11-27', 0);

INSERT INTO Emprestimos (id_usuario, id_livro, data_emprestimo, data_prevista_devolucao, renovacoes)
VALUES (2, 2, '2024-11-19', '2024-11-26', 0);
```
