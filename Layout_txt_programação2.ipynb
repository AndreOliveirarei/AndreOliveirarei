```python

with open('arquivo.txt', 'r') as arquivo:
    conteudo = arquivo.read().splitlines()

for transacao in conteudo:
    id_transacao = transacao[0:10]
    tipo_transacao = transacao[10:11]
    data_transacao = transacao[11:19]
    hora_transacao = transacao[19:25]
    valor_transacao = transacao[25:35]
    conta_origem = transacao[35:45]
    conta_destino = transacao[45:55]

    # Extraindo dia, mês e ano da data
    dia = data_transacao[0:2]
    mes = data_transacao[2:4]
    ano = data_transacao[4:8]

    # Imprimindo os detalhes
    print(f"ID da Transação: {id_transacao}")
    print(f"Tipo de Transação: {tipo_transacao}")
    print(f"Data da Transação: {dia}/{mes}/{ano}")
    print(f"Hora da Transação: {hora_transacao}")
    print(f"Valor da Transação: {valor_transacao}")
    print(f"Conta de Origem: {conta_origem}")
    print(f"Conta de Destino: {conta_destino}")
    print("-" * 30)


0000000001C202401151030000000150001234567890987654321
0000000002D202401151035000000089509876543211234567890
0000000003C202401151042000001200005555666667777888899
```
