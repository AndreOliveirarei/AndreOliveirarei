```python
with open('arquivo.txt', 'r') as arquivo:
    conteudo = arquivo.read().splitlines()

for data in conteudo:
    dia = data[0:2]
    mes = data[2:4]
    ano = data[4:8]

    print(f"O dia é: {dia}")
    print(f"O mês é: {mes}")
    print(f"O ano é: {ano}")
    print("-" * 30)
```
